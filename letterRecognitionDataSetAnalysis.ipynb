```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
from sklearn.metrics import classification_report, confusion_matrix, roc_curve, auc
from sklearn.preprocessing import label_binarize
print("libraries loaded")
```

```python
file="/content/letterRecogniton/letter-recognition.data"
df = pd.read_csv(file, header=None)
df.head()
```

```python
column_names = [
    'lettr', 'x-box', 'y-box', 'width', 'high', 'onpix',
    'x-bar', 'y-bar', 'x2bar', 'y2bar', 'xybar', 'x2ybr',
    'xy2br','x-ege', 'xegvy', 'y-ege', 'yegvx'
]
df.columns = column_names
df.head()
```

```python
df.describe(include='all')
```

```python
df.isnull().sum()
```

```python
# harf sütunuyla özellik sütunları ayrılıyor
X = df.iloc[:, 1:] # özellikler
Y = df.iloc[:, 0] # harfler

# normalizasyon
scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)

# veri bölmw
X_train, X_test, Y_train, Y_test = train_test_split(X_scaled, Y, test_size=0.20, random_state=42)
# k değeri aranıyor
success_rates = []
print("k values are being tested...")
for k in range(1, 21):
  knn = KNeighborsClassifier(n_neighbors=k)
  knn.fit(X_train, Y_train) # eğitim

  predictions = knn.predict(X_test)
  accuracy = accuracy_score(Y_test, predictions)
  success_rates.append(accuracy)
  print(f"Accuracy for K Value = {k} %{accuracy*100:.2f}")

plt.figure(figsize=(10, 6))
plt.plot(range(1, 21), success_rates, color='blue', linestyle='dashed', marker='o', markerfacecolor='red', markersize=8)
plt.title('k Value Accuracy')
plt.xlabel('K (Neighbors Count))')
plt.ylabel('Accuracy')
plt.xticks(range(1, 21))
plt.grid(True)
plt.show()
```

```python
# overfitting var mı yok mu
knn1 = KNeighborsClassifier(n_neighbors=1)
knn1.fit(X_train, Y_train)

train_acc = accuracy_score(Y_train, knn1.predict(X_train))
test_acc = accuracy_score(Y_test, knn1.predict(X_test))

print(f"Train Accuracy: %{train_acc*100:.2f}")
print(f"Test Accuracy:  %{test_acc*100:.2f}")
```

```python
# k=5
final_knn = KNeighborsClassifier(n_neighbors=5)
final_knn.fit(X_train, Y_train)

Y_pred = final_knn.predict(X_test)
print("----- RESULTS -----")

letters = sorted(list(Y_train.unique()))
print(classification_report(Y_test, Y_pred, target_names=letters))
```

```python
# confusion matrix
cm = confusion_matrix(Y_test, Y_pred)
plt.figure(figsize=(15, 12))

sns.heatmap(cm, annot=True, fmt='d', cmap='YlGnBu',
            xticklabels=letters, yticklabels=letters,
            cbar_kws={'label': 'Prediction Count'})
plt.title('Letter Recognition Model - Confusion Matrix', fontsize=16, fontweight='bold')
plt.ylabel('Actual Label', fontsize=12)
plt.xlabel('Predicted Label', fontsize=12)
plt.tight_layout()
plt.show()

# --- Class-Based F1-Score Bar Chart ---
rapor_dict = classification_report(Y_test, Y_pred, target_names=letters, output_dict=True)
f1_scores = [rapor_dict[letter]['f1-score'] for letter in letters]

plt.figure(figsize=(14, 6))
colors = sns.color_palette("RdYlGn", len(letters))
sequential_indexes = np.argsort(f1_scores)
bar_colors = [None] * len(letters)
for rank, idx in enumerate(sequential_indexes):
    bar_colors[idx] = colors[rank]

bars = plt.bar(letters, f1_scores, color=bar_colors, edgecolor='black', alpha=0.8)
plt.axhline(y=np.mean(f1_scores), color='red', linestyle='--', label=f'Avarage F1-Score: %{np.mean(f1_scores)*100:.1f}')

plt.title('Label Based F1-Score Success Levels', fontsize=16, fontweight='bold')
plt.xlabel('Letters', fontsize=12)
plt.ylabel('F1-Score', fontsize=12)
plt.ylim(0.80, 1.02)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.legend()
plt.tight_layout()
plt.show()

# --- ROC / AUC CURVE ---
y_test_bin = label_binarize(Y_test, classes=letters) #binarzation
y_score = final_knn.predict_proba(X_test)

plt.figure(figsize=(12, 8))

fpr = dict()
tpr = dict()
roc_auc = dict()

selected_letters = ['A', 'B', 'H', 'K', 'W', 'Z'] # W->best H->worst prediction

for i in letters:
    idx = letters.index(i)
    fpr[i], tpr[i], _ = roc_curve(y_test_bin[:, idx], y_score[:, idx])
    roc_auc[i] = auc(fpr[i], tpr[i])

for i in selected_letters:
    plt.plot(fpr[i], tpr[i], label=f'Letter {i} (AUC = {roc_auc[i]:.3f})')

plt.plot([0, 1], [0, 1], 'k--', linestyle=':', label='Random Prediction (AUC = 0.50)')

plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate', fontsize=12)
plt.ylabel('True Positive Rate', fontsize=12)
plt.title('Multi-class ROC Curve', fontsize=16, fontweight='bold')
plt.legend(loc="lower right")
plt.grid(True)
plt.tight_layout()
plt.show()
```
