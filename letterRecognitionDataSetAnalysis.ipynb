```python
file="/content/letterRecogniton/letter-recognition.data"
df = pd.read_csv(file, header=None)
df.head()
```

```python
column_names = [
    'lettr', 'x-box', 'y-box', 'width', 'high', 'onpix',
    'x-bar', 'y-bar', 'x2bar', 'y2bar', 'xybar', 'x2ybr',
    'xy2br','x-ege', 'xegvy', 'y-ege', 'yegvx'
]
df.columns = column_names
df.head()
```

```python
df.describe(include='all')
```

```python
df.isnull().sum()
```

```python
# harf sütunuyla özellik sütunları ayrılıyor
X = df.iloc[:, 1:] # özellikler
Y = df.iloc[:, 0] # harfler

# normalizasyon
scaler = MinMaxScaler()
X_scaled = scaler.fit_transform(X)

# veri bölmw
X_train, X_test, Y_train, Y_test = train_test_split(X_scaled, Y, test_size=0.20, random_state=42)
# k değeri aranıyor
success_rates = []
print("k values are being tested...")
for k in range(1, 21):
  knn = KNeighborsClassifier(n_neighbors=k)
  knn.fit(X_train, Y_train) # eğitim

  predictions = knn.predict(X_test)
  accuracy = accuracy_score(Y_test, predictions)
  success_rates.append(accuracy)
  print(f"Accuracy for K Value = {k} %{accuracy*100:.2f}")

plt.figure(figsize=(10, 6))
plt.plot(range(1, 21), success_rates, color='blue', linestyle='dashed', marker='o', markerfacecolor='red', markersize=8)
plt.title('k Value Accuracy')
plt.xlabel('K (Neighbors Count))')
plt.ylabel('Accuracy')
plt.xticks(range(1, 21))
plt.grid(True)
plt.show()
```

```python
# overfitting var mı yok mu
knn1 = KNeighborsClassifier(n_neighbors=1)
knn1.fit(X_train, Y_train)

train_acc = accuracy_score(Y_train, knn1.predict(X_train))
test_acc = accuracy_score(Y_test, knn1.predict(X_test))

print(f"Train Accuracy: %{train_acc*100:.2f}")
print(f"Test Accuracy:  %{test_acc*100:.2f}")
```
